# EDA y Predicción de Precios de Airbnb

Proyecto de análisis + regresión listo para **Google Colab**.

**Pregunta:** ¿qué factores influyen en el precio de un listing de Airbnb?

**Target:** `price`

**Stack:** pandas, seaborn, scikit-learn, shap

Dataset de ejemplo (proxy de Airbnb).

## 0. Setup

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
print("OK")
```

## 1. Cargar datos

```python
url = "https://raw.githubusercontent.com/plotly/datasets/master/tips.csv"
df = pd.read_csv(url)
df["price"] = df["total_bill"] * 10
print(df.shape)
df.head()
```

## 2. EDA

```python
print(df.describe().T)
print("\nNulos:", df.isnull().sum().sum())
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df["price"], kde=True, bins=30)
plt.title("Distribución de precios")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x="day", y="price", ax=axes[0])
axes[0].set_title("Precio por día")
sns.boxplot(data=df, x="time", y="price", ax=axes[1])
axes[1].set_title("Precio por momento")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.scatterplot(data=df, x="total_bill", y="price", hue="size")
plt.title("Total bill vs Precio")
plt.show()
```

### Hallazgos
- El precio depende directamente de variables numéricas base.
- Diferencias por día y momento.
- Ideal para practicar EDA de precios y categóricas.

## 3. Preparar datos

```python
X = df.drop(columns=["price"])
y = df["price"]

cat_cols = X.select_dtypes(include="object").columns.tolist()
X = pd.get_dummies(X, columns=cat_cols, drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)
print(X_train.shape, X_test.shape)
```

## 4. Modelos

```python
modelos = {
    "Ridge": Ridge(random_state=RANDOM_STATE),
    "Random Forest": RandomForestRegressor(n_estimators=200, random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
}

resultados = []
pipes = {}

for nombre, clf in modelos.items():
    pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", clf)
    ])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    resultados.append({
        "modelo": nombre,
        "MAE": mean_absolute_error(y_test, y_pred),
        "RMSE": mean_squared_error(y_test, y_pred, squared=False),
        "R2": r2_score(y_test, y_pred),
    })
    pipes[nombre] = pipe

pd.DataFrame(resultados).sort_values("R2", ascending=False)
```

## 5. Feature importance

```python
rf = pipes["Random Forest"].named_steps["model"]
importancias = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False).head(10)

plt.figure(figsize=(8, 4))
sns.barplot(x=importancias.values, y=importancias.index)
plt.title("Top features — Random Forest")
plt.tight_layout()
plt.show()
```

## 6. SHAP

```python
import shap

rf_pipe = pipes["Random Forest"]
X_train_trans = rf_pipe.named_steps["imputer"].transform(X_train)
X_train_trans = rf_pipe.named_steps["scaler"].transform(X_train_trans)
model = rf_pipe.named_steps["model"]

explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_train_trans)

print("SHAP listo")
```

```python
shap.summary_plot(shap_values, X_train_trans, feature_names=X.columns, max_display=8, show=False)
plt.title("SHAP Summary — Precios")
plt.tight_layout()
plt.show()
```

```python
shap.summary_plot(shap_values, X_train_trans, feature_names=X.columns, plot_type="bar", max_display=8, show=False)
plt.title("Importancia media SHAP")
plt.tight_layout()
plt.show()
```

## 7. Conclusiones

- El precio depende fuertemente de variables numéricas base.
- R² suele ser alto en este ejemplo.
